````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 및 마이너스 기호 깨짐 방지
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.unicode_minus'] = False

# 원본 데이터 로드 (원본 파일은 절대 수정하지 않고 읽기만 합니다)
df = pd.read_csv('diabetes_binary_health_indicators_BRFSS2015.csv')

# 1. 기본 정보 확인
print(f"데이터 크기 (행, 열): {df.shape}")
display(df.head(3))

# 2. 결측치 및 데이터 타입 확인
print("\n--- 결측치 확인 ---")
print(df.isnull().sum())
````

````python
# 타겟 변수 분포 건수 및 비율 확인
target_counts = df['Diabetes_binary'].value_counts()
target_ratio = df['Diabetes_binary'].value_counts(normalize=True) * 100

print("--- 당뇨 여부 분포 (건수) ---")
print(target_counts)
print("\n--- 당뇨 여부 분포 (%) ---")
print(target_ratio)

# 시각화로 확인
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Diabetes_binary', palette='Set2')
plt.title('Distribution of Diabetes Binary (0: Normal, 1: Diabetes)')
plt.xlabel('Diabetes Status')
plt.ylabel('Count')
plt.show()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 타겟 변수(Diabetes_binary) 분포 건수 및 비율 계산
target_counts = df['Diabetes_binary'].value_counts().reset_index()
target_counts.columns = ['Diabetes_status', 'Count']
target_counts['Ratio(%)'] = (target_counts['Count'] / len(df)) * 100

print("--- 당뇨 여부 분포 요약 ---")
display(target_counts)

# [자동 저장 명령어 1] 집계된 통계 데이터를 interim 폴더에 CSV로 저장
target_counts.to_csv('data/interim/diabetes_target_distribution.csv', index=False, encoding='utf-8-sig')

# 2. 타겟 변수 분포 시각화 및 이미지 저장
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Diabetes_binary', palette='Set2')
plt.title('Distribution of Diabetes Binary (0: Normal, 1: Diabetes)')
plt.xlabel('Diabetes Status')
plt.ylabel('Count')

# [자동 저장 명령어 2] 시각화 그래프를 reports/figures 폴더에 이미지로 저장
plt.savefig('reports/figures/diabetes_distribution.png', bbox_inches='tight')

# 그래프 화면에 출력
plt.show()
````
