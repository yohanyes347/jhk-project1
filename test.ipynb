````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 0. 환경 설정 및 폰트 깨짐 방지
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.unicode_minus'] = False

# interim 하위 폴더 생성 (혼선 방지를 위해 분석 주제별로 폴더 분기)
os.makedirs('data/interim/sec1_demographics', exist_ok=True)
os.makedirs('data/interim/sec2_metabolic_interaction', exist_ok=True)
os.makedirs('data/interim/sec3_lifestyle_cluster', exist_ok=True)
os.makedirs('data/interim/sec4_socioeconomic', exist_ok=True)
os.makedirs('reports/figures', exist_ok=True)

# 원본 데이터 로드 (notebooks 폴더 안에서 실행하는 경우 상위 경로 지정, 필요시 조정)
# df = pd.read_csv('../diabetes_binary_health_indicators_BRFSS2015.csv')
df = pd.read_csv('diabetes_binary_health_indicators_BRFSS2015.csv')
print(f"데이터 로드 완료! 전체 크기: {df.shape}")
````

## [섹션 1: 인구통계학적 요인 분석 요약]
- 범례 및 그룹 정의: 초록색(0.0)은 여성, 주황색(1.0)은 남성을 나타내며, 연령은 1그룹(18~24세), 2그룹(25-29세)부터 시작해 13그룹(80세 이상)까지 세분화된 구간으로 분석됨.
- 핵심 인사이트: 본 데이터셋에서 연령(Age)은 당뇨 발병에 가장 강력한 선행 지표 중 하나로 작용하며, 특히 남성 집단이 여성 집단에 비해 전반적인 연령대에서 더 높은 유병 리스크를 안고 있음을 통계적으로 확인했음.


````python
print("\n--- [Section 2] 대사질환 복합 요인 상호작용 분석 시작 ---")

# 복합 위험군 파생변수 생성 (고혈압 * 고콜레스테롤, 비만 여부 추가)
df['Obese'] = (df['BMI'] >= 30).astype(int)
df['BP_Chol_Interaction'] = df['HighBP'] * df['HighChol']
df['Triple_Risk'] = df['HighBP'] * df['HighChol'] * df['Obese']

# 그룹별 당뇨 발병률 집계
sec2_df = df.groupby('Triple_Risk')['Diabetes_binary'].agg(['count', 'mean']).reset_index()
sec2_df.columns = ['Triple_Risk_Condition', 'Total_Count', 'Diabetes_Probability']

# interim 폴더 하위 sec2에 저장
sec2_path = 'data/interim/sec2_metabolic_interaction/metabolic_synergy_rate.csv'
sec2_df.to_csv(sec2_path, index=False, encoding='utf-8-sig')
print(f"-> 저장 완료: {sec2_path}")

# 시각화 및 저장
plt.figure(figsize=(8, 5))
sns.barplot(data=sec2_df, x='Triple_Risk_Condition', y='Diabetes_Probability', palette='coolwarm')
plt.title('Diabetes Probability by Triple Metabolic Risk (BP x Chol x Obesity)')
plt.xlabel('Triple Risk Condition (0: No, 1: Yes)')
plt.ylabel('Diabetes Probability')
plt.savefig('reports/figures/sec2_triple_risk.png', bbox_inches='tight')
plt.show()
````

## [섹션 2: 대사질환 복합 요인 상호작용 분석 요약]

분석 내용: 고혈압(HighBP), 고콜레스테롤(HighChol), 비만(BMI≥30)의 삼중 위험 요인 상호작용(Triple_Risk)에 따른 당뇨 발병 확률 비교.

핵심 인사이트: 3가지 위험 요인이 모두 겹친 고위험군 집단(1.0)의 당뇨 발병 확률은 약 39%로, 비위험군(0.0, 약 10.5%) 대비 약 3.7배 이상 급증함. 이는 대사 질환 간의 복합적 상호작용이 당뇨 발병

````python
print("\n--- [Section 3] 생활 습관 고위험군 군집 분석 시작 ---")

# 최악의 생활습관 플래그 생성 (운동 안 함(PhysActivity==0) & 흡연자(Smoker==1))
df['Bad_Lifestyle'] = ((df['PhysActivity'] == 0) & (df['Smoker'] == 1)).astype(int)

sec3_df = df.groupby('Bad_Lifestyle')['Diabetes_binary'].agg(['count', 'mean']).reset_index()
sec3_df.columns = ['Bad_Lifestyle_Cluster', 'Total_Count', 'Diabetes_Probability']

# interim 폴더 하위 sec3에 저장
sec3_path = 'data/interim/sec3_lifestyle_cluster/lifestyle_cluster_rate.csv'
sec3_df.to_csv(sec3_path, index=False, encoding='utf-8-sig')
print(f"-> 저장 완료: {sec3_path}")

# 시각화 및 저장
plt.figure(figsize=(7, 5))
sns.barplot(data=sec3_df, x='Bad_Lifestyle_Cluster', y='Diabetes_Probability', palette='pastel')
plt.title('Diabetes Probability by Bad Lifestyle Cluster (No Exercise & Smoker)')
plt.xlabel('Bad Lifestyle Cluster (0: Normal, 1: High Risk)')
plt.ylabel('Diabetes Probability')
plt.savefig('reports/figures/sec3_bad_lifestyle.png', bbox_inches='tight')
plt.show()
````

## [섹션 3: 생활 습관 고위험군 군집 분석 요약]

분석 내용: 운동 부족(PhysActivity == 0)과 흡연(Smoker == 1)이 결합된 생활 습관 고위험군 군집(Bad_Lifestyle)에 따른 당뇨 발병 확률 비교.

핵심 인사이트: 운동 부족과 흡연이 겹친 고위험군 집단(1.0)의 당뇨 발병 확률은 약 22.3%로, 일반 집단(0.0, 약 12.8%) 대비 유병률이 대폭 상승함. 이는 생활 습관 불량 요소가 당뇨 발병에 직접적인 위험 요인으로 작용함을 증명함.

````python
print("\n--- [Section 4] 사회경제적 여건 및 건강 상태 격차 분석 시작 ---")

# 소득(Income) 및 교육(Education) 수준에 따른 평균 건강 상태(GenHlth) 및 당뇨 유병률 교차 분석
sec4_df = df.groupby(['Income', 'Education'])[['GenHlth', 'Diabetes_binary']].mean().reset_index()

# interim 폴더 하위 sec4에 저장
sec4_path = 'data/interim/sec4_socioeconomic/income_education_health_matrix.csv'
sec4_df.to_csv(sec4_path, index=False, encoding='utf-8-sig')
print(f"-> 저장 완료: {sec4_path}")

# 시각화 및 저장 (소득 수준별 당뇨 유병률 추이)
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='Income', y='Diabetes_binary', hue='Education', marker='o', palette='viridis')
plt.title('Diabetes Probability by Income and Education Level')
plt.xlabel('Income Level')
plt.ylabel('Diabetes Probability')
plt.legend(title='Education', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.savefig('reports/figures/sec4_socioeconomic_trend.png', bbox_inches='tight')
plt.show()

print("\n🎉 모든 섹션 분석 및 interim/reports 폴더 정리 완료!")
````

## [섹션 4: 사회경제적 여건 및 건강 상태 격차 분석 요약]

분석 내용: 소득 수준(Income)과 교육 수준(Education)에 따른 당뇨 발병 확률 교차 분석 및 추이 시각화.

핵심 인사이트: 소득 및 교육 수준이 낮을수록 당뇨 발병 확률이 급격히 상승하며(최대 40% 임박), 반대로 상위 계층으로 갈수록 리스크가 대폭 완화됨을 확인. 이는 경제·교육적 여건이 건강 상태 및 질병 발병에 강력한 간접 결정 요인으로 작용함을 시사함.

## 범례: 
가. 교육 수준 (Education) 범주 (1 ~ 6단계)
설문 대상자가 완료한 최종 학력을 기준(서열 척도)으로 6개 구간으로 나뉩니다:

1.0: 무학 또는 유치원까지만 수료 (Never attended school / Kindergarten only)

2.0: 초등학교 졸업 또는 중학교 과정 (Elementary / Grades 1 through 8)

3.0: 고등학교 중퇴 (Grades 9–11, Some high school)

4.0: 고등학교 졸업 또는 검정고시(GED) (High school graduate / GED)

5.0: 대학교 재학/휴학 또는 기술전문대 1~3년 수료 (Some college or technical school)

6.0: 대학교 졸업 이상 (4 years or more of college / College graduate)


나. 소득 수준 (Income Level) 범주 (1 ~ 8단계)
가구의 모든 소득원을 합산한 연간 총 가구 소득(Annual Household Income)을 기준으로 8개 구간으로 나뉩니다:

1: 10,000 미만 (Less than $10,000)

2: $10,000 이상 ~ $15,000 미만

3: $15,000 이상 ~ $20,000 미만

4: $20,000 이상 ~ $25,000 미만

5: $25,000 이상 ~ $35,000 미만

2. 핵심 변수 심층 해석

### ① Triple_Risk (대사질환 삼중 위험군)
* **오즈비 (OR 4.987):** 모델 내에서 가장 파괴적인 영향력을 행사하는 핵심 위험 인자
* **해석:** 고혈압, 고콜레스테롤, 비만을 모두 동반한 경우 당뇨병 발병 위험이 약 5배 급증하므로, 1차 예방 보건 정책의 최우선 타겟군으로 설정해야 함

### ② Bad_Lifestyle (생활습관 고위험군)
* **오즈비 (OR 1.664):** 운동 부족과 흡연이라는 후천적 악습의 독립적 위험성 입증
* **해석:** 대사질환이 없더라도 불건전한 생활 습관 자체만으로 당뇨 위험을 1.66배 높임

### ③ Super_High_Risk (최종 복합 교차 고위험군)
* **오즈비 (OR 0.809, p < 0.001):** 
  * 1 미만의 수치로 나타난 이유는 **포화(Saturation) 효과**에 기인함
  * 이미 `Triple_Risk`(OR ~4.98)라는 압도적인 개별 위험이 반영된 상태에서, 생활습관 악화가 겹칠 때 추가 위험의 기울기가 완만해지는 통계적 보정 현상임
  * 그럼에도 **p-value가 $1.11 \times 10^{-9}$로 극도로 유의미**하므로, 최악의 복합 고위험 통제선으로서 강력한 통계적 타당성을 확보함

## 결측치 64개는 중앙값을 넣고, 이상치 584개는 제거하는 방식으로 수정.  정제된 데이터는 'data/processed' 폴더에 보관

````python
import pandas as pd
import os

# 1. 원본 데이터 불러오기
file_path = 'data/raw/diabetes_binary_health_indicators_BRFSS2015.csv'
# 만약 파일이 현재 경로에 바로 있다면 아래 주석처럼 사용하세요
# file_path = 'diabetes_binary_health_indicators_BRFSS2015.csv'

df = pd.read_csv(file_path)
print(f"📌 원본 데이터 총 행 수: {len(df):,}개")

# 2. 결측치 처리: 수치형/전체 변수의 시스템 결측치(NaN)를 '중앙값(Median)'으로 대체
print(f"🔍 처리 전 결측치 총합: {df.isnull().sum().sum()}개")

# 수치형 컬럼들의 결측치를 각 컬럼의 중앙값으로 채우기
# (문자열이나 범주형이 포함되어 있다면 수치형 컬럼만 골라서 채울 수도 있습니다)
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"  - '{col}' 결측치를 중앙값({median_val})으로 대체했습니다.")

print(f"✨ 결측치 처리 후 총합: {df.isnull().sum().sum()}개")


# 3. 이상치 처리: 비정상적인 BMI 및 나이(Age) 범위를 가진 행(Row) 제거
initial_len = len(df)

# - BMI 정상 범위: 10 초과 ~ 60 미만 (극단적 오류 값 제거)
# - Age 정상 범위: BRFSS 코드 기준 1 ~ 13
df_cleaned = df[
    (df['BMI'] > 10) & (df['BMI'] < 60) & 
    (df['Age'] >= 1) & (df['Age'] <= 13)
].copy()

removed_count = initial_len - len(df_cleaned)
print(f"🗑️ 제거된 이상치 행 수: {removed_count:,}개")
print(f"✨ 정제 후 최종 데이터 총 행 수: {len(df_cleaned):,}개")


# 4. 'data/processed' 디렉토리가 없다면 자동 생성 후 CSV 저장
output_dir = 'data/processed'
os.makedirs(output_dir, exist_ok=True)

output_file_path = os.path.join(output_dir, 'diabetes_binary_health_indicators_cleaned.csv')
df_cleaned.to_csv(output_file_path, index=False)

print(f"💾 정제된 고품질 데이터 저장 완료: '{output_file_path}'")
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 정제된 데이터 로드 (processed 폴더 활용)
processed_file_path = 'data/processed/diabetes_binary_health_indicators_cleaned.csv'
df = pd.read_csv(processed_file_path)

print(f"📌 데이터 로드 완료: 총 {len(df):,}개 행, {df.shape[1]}개 컬럼")

# 2. 주요 수치형 및 설문 변수 선택 (데이터셋에 존재하는 컬럼만 필터링)
target_cols = [
    'Diabetes_binary', 'HighBP', 'HighChol', 'BMI', 'Age', 
    'MentHlth', 'PhysHlth', 'GenHlth', 'Education', 'Income'
]
selected_cols = [col for col in target_cols if col in df.columns]

corr_df = df[selected_cols]

# 3. 상관관계 행렬(Correlation Matrix) 계산
corr_matrix = corr_df.corr(method='pearson')

# 4. 상관관계 히트맵 시각화
plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix, 
    annot=True, 
    fmt='.2f', 
    cmap='coolwarm', 
    vmin=-1, 
    vmax=1, 
    linewidths=0.5,
    cbar_kws={'label': 'Pearson Correlation Coefficient'}
)
plt.title('Correlation Heatmap of Key Features', fontsize=14, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()

# 5. 상관도가 높은(절대값 기준 0.5 이상 등) 변수 쌍 탐색 출력
print("=== [참고] 상관도가 높은 주요 변수 쌍 (Absolute Corr > 0.4) ===")
high_corr_pairs = []
for i in range(len(corr_matrix.columns)):
    for j in range(i):
        corr_val = corr_matrix.iloc[i, j]
        if abs(corr_val) > 0.4:
            col1 = corr_matrix.columns[i]
            col2 = corr_matrix.columns[j]
            high_corr_pairs.append((col1, col2, corr_val))

if high_corr_pairs:
    for c1, c2, val in high_corr_pairs:
        print(f"• {c1} & {c2}: 상관계수 = {val:.4f}")
else:
    print("기준치(0.4)를 넘는 고도 상관 쌍이 없습니다.")
````

````python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 한글 폰트 깨짐 방지 (시스템 환경에 따라 설정)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 정제된 데이터 로드
processed_file_path = 'data/processed/diabetes_binary_health_indicators_cleaned.csv'
df = pd.read_csv(processed_file_path)

print("📌 데이터 로드 완료. sec1 ~ sec6 폴더별 도표 생성 및 저장을 시작합니다...\n")

# ==========================================
# [Sec 1] 탐색적 데이터 분석 (EDA) - 타겟 분포
# ==========================================
sec1_dir = 'data/interim/sec1'
os.makedirs(sec1_dir, exist_ok=True)

plt.figure(figsize=(6, 4))
ax1 = sns.countplot(x='Diabetes_binary', hue='Diabetes_binary', data=df, palette=['#4C72B0', '#DD8452'], legend=False)
plt.title('Sec 1: 타겟 변수(Diabetes_binary) 분포', fontsize=12, fontweight='bold')
plt.xlabel('Class (0: Normal, 1: Diabetes)')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig(os.path.join(sec1_dir, 'class_distribution.png'), dpi=300)
plt.close()
print("📁 [Sec 1] 저장 완료: data/interim/sec1/class_distribution.png")


# ==========================================
# [Sec 2] 주요 수치형 변수 분포 (BMI, Age)
# ==========================================
sec2_dir = 'data/interim/sec2'
os.makedirs(sec2_dir, exist_ok=True)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df['BMI'], bins=30, ax=axes[0], color='#4C72B0', kde=True)
axes[0].set_title('Sec 2: BMI 분포 정제 상태', fontweight='bold')

sns.countplot(x='Age', hue='Age', data=df, ax=axes[1], palette='viridis', legend=False)
axes[1].set_title('Sec 2: Age(연령대) 분포', fontweight='bold')

plt.tight_layout()
plt.savefig(os.path.join(sec2_dir, 'bmi_age_distribution.png'), dpi=300)
plt.close()
print("📁 [Sec 2] 저장 완료: data/interim/sec2/bmi_age_distribution.png")


# ==========================================
# [Sec 3] 건강 상태 및 생활습관 변수 분포
# ==========================================
sec3_dir = 'data/interim/sec3'
os.makedirs(sec3_dir, exist_ok=True)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(x='GenHlth', hue='GenHlth', data=df, ax=axes[0], palette='Set2', legend=False)
axes[0].set_title('Sec 3: 일반 건강 상태 (GenHlth)', fontweight='bold')

sns.countplot(x='HighBP', hue='HighBP', data=df, ax=axes[1], palette='coolwarm', legend=False)
axes[1].set_title('Sec 3: 고혈압 보유 여부 (HighBP)', fontweight='bold')

plt.tight_layout()
plt.savefig(os.path.join(sec3_dir, 'health_status_distribution.png'), dpi=300)
plt.close()
print("📁 [Sec 3] 저장 완료: data/interim/sec3/health_status_distribution.png")


# ==========================================
# [Sec 4] 클래스 불균형 시각화 (비율 비교)
# ==========================================
sec4_dir = 'data/interim/sec4'
os.makedirs(sec4_dir, exist_ok=True)

plt.figure(figsize=(5, 5))
counts = df['Diabetes_binary'].value_counts()
plt.pie(counts, labels=['정상인 (0)', '당뇨 환자 (1)'], autopct='%.1f%%', startangle=90, colors=['#4C72B0', '#DD8452'])
plt.title('Sec 4: 클래스 불균형 비율 (Pie Chart)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(sec4_dir, 'class_imbalance_pie.png'), dpi=300)
plt.close()
print("📁 [Sec 4] 저장 완료: data/interim/sec4/class_imbalance_pie.png")


# ==========================================
# [Sec 5] 데이터 누수 차단 전제 주요 피처 박스플롯 (BMI vs Diabetes)
# ==========================================
sec5_dir = 'data/interim/sec5'
os.makedirs(sec5_dir, exist_ok=True)

plt.figure(figsize=(6, 4))
sns.boxplot(x='Diabetes_binary', y='BMI', hue='Diabetes_binary', data=df, palette=['#4C72B0', '#DD8452'], legend=False)
plt.title('Sec 5: 당뇨 여부에 따른 BMI 비교', fontsize=12, fontweight='bold')
plt.xlabel('Diabetes_binary')
plt.ylabel('BMI')
plt.tight_layout()
plt.savefig(os.path.join(sec5_dir, 'bmi_by_diabetes_boxplot.png'), dpi=300)
plt.close()
print("📁 [Sec 5] 저장 완료: data/interim/sec5/bmi_by_diabetes_boxplot.png")


# ==========================================
# [Sec 6] 다중공선성 검토 (상관관계 히트맵)
# ==========================================
sec6_dir = 'data/interim/sec6'
os.makedirs(sec6_dir, exist_ok=True)

target_cols = [
    'Diabetes_binary', 'HighBP', 'HighChol', 'BMI', 'Age', 
    'MentHlth', 'PhysHlth', 'GenHlth', 'Education', 'Income'
]
corr_matrix = df[[col for col in target_cols if col in df.columns]].corr(method='pearson')

plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix, 
    annot=True, 
    fmt='.2f', 
    cmap='coolwarm', 
    vmin=-1, 
    vmax=1, 
    linewidths=0.5,
    cbar_kws={'label': 'Pearson Correlation'}
)
plt.title('Sec 6: 주요 변수 간 상관관계 히트맵', fontsize=14, fontweight='bold', pad=15)
plt.tight_layout()
plt.savefig(os.path.join(sec6_dir, 'correlation_heatmap.png'), dpi=300)
plt.close()
print("📁 [Sec 6] 저장 완료: data/interim/sec6/correlation_heatmap.png")

print("\n✨ 모든 구간(Sec 1 ~ Sec 6)의 도표 자료가 해당 폴더에 성공적으로 저장되었습니다!")
````
